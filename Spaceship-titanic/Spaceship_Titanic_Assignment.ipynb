```python
# ============================================================
# DATA PREPROCESSING
# ============================================================

train = train_df.copy()
test = test_df.copy()

# ------------------------------------------------------------
# Feature Extraction
# ------------------------------------------------------------

for df in [train, test]:

    # Cabin Features
    df[['Deck', 'CabinNum', 'Side']] = df['Cabin'].str.split('/', expand=True)
    df['CabinNum'] = pd.to_numeric(df['CabinNum'], errors='coerce')

    # Passenger Group Features
    df[['GroupID', 'GroupMember']] = df['PassengerId'].str.split('_', expand=True)

    df['GroupID'] = df['GroupID'].astype(int)
    df['GroupMember'] = df['GroupMember'].astype(int)

    df['GroupSize'] = df.groupby('GroupID')['GroupID'].transform('count')

# ------------------------------------------------------------
# Spending Features
# ------------------------------------------------------------

spending_cols = [
    'RoomService',
    'FoodCourt',
    'ShoppingMall',
    'Spa',
    'VRDeck'
]

# Missing spending means no spending
for df in [train, test]:
    df[spending_cols] = df[spending_cols].fillna(0)

for df in [train, test]:

    df["TotalSpend"] = df[spending_cols].sum(axis=1)

    df["HasSpend"] = (df["TotalSpend"] > 0).astype(int)

    df["NoSpending"] = (df["TotalSpend"] == 0).astype(int)

    df["LogSpend"] = np.log1p(df["TotalSpend"])
    # 1. SpendPerPerson & 3. LuxurySpend
    df["SpendPerPerson"] = df["TotalSpend"] / df["GroupSize"]
    df["LuxurySpend"] = df["Spa"] + df["VRDeck"]
    df["EssentialSpend"] = (
        df["FoodCourt"] +
        df["ShoppingMall"]
    )
    df["LuxuryRatio"] = (
        df["LuxurySpend"] /
        (df["TotalSpend"] + 1)
    )

# ------------------------------------------------------------
# Family Features
# ------------------------------------------------------------
# 2. PairTravel & SmallGroup
for df in [train, test]:

    df["IsAlone"] = (df["GroupSize"] == 1).astype(int)

    df["IsFamily"] = (df["GroupSize"] > 1).astype(int)

    df["PairTravel"] = (df["GroupSize"] == 2).astype(int)

    df["SmallGroup"] = (
        (df["GroupSize"] >= 2) &
        (df["GroupSize"] <= 4)
    ).astype(int)

    df["LargeGroup"] = (df["GroupSize"] >= 5).astype(int)

# ------------------------------------------------------------
# HomePlanet Imputation (Group Wise)
# ------------------------------------------------------------

for df in [train, test]:

    df["HomePlanet"] = (
        df.groupby("GroupID")["HomePlanet"]
        .transform(lambda x: x.fillna(x.mode().iloc[0]) if not x.mode().empty else x)
    )

mode_home = train["HomePlanet"].mode()[0]

train["HomePlanet"].fillna(mode_home, inplace=True)
test["HomePlanet"].fillna(mode_home, inplace=True)

# Suggested Feature Addition: PlanetDestination
for df in [train, test]:
    df["PlanetDestination"] = (
        df["HomePlanet"].astype(str) + "_" + df["Destination"].astype(str)
    )

# ------------------------------------------------------------
# Destination Imputation (Group Wise)
# ------------------------------------------------------------

for df in [train, test]:

    df["Destination"] = (
        df.groupby("GroupID")["Destination"]
        .transform(lambda x: x.fillna(x.mode().iloc[0]) if not x.mode().empty else x)
    )

mode_dest = train["Destination"].mode()[0]

train["Destination"].fillna(mode_dest, inplace=True)
test["Destination"].fillna(mode_dest, inplace=True)

# ------------------------------------------------------------
# Age Imputation (HomePlanet + Deck Median)
# ------------------------------------------------------------

train["Age"] = train.groupby(
    ["HomePlanet", "Deck"]
)["Age"].transform(lambda x: x.fillna(x.median()))

test["Age"] = test.groupby(
    ["HomePlanet", "Deck"]
)["Age"].transform(lambda x: x.fillna(x.median()))

median_age = train["Age"].median()

train["Age"].fillna(median_age, inplace=True)
test["Age"].fillna(median_age, inplace=True)

# ------------------------------------------------------------
# Age Groups
# ------------------------------------------------------------
# 6. Better Age Groups
bins = [0,5,12,18,25,40,60,100]

labels = [
    "Infant",
    "Child",
    "Teen",
    "YoungAdult",
    "Adult",
    "MiddleAge",
    "Senior"
]

train["AgeGroup"] = pd.cut(train["Age"], bins=bins, labels=labels, include_lowest=True)

test["AgeGroup"] = pd.cut(test["Age"], bins=bins, labels=labels, include_lowest=True)

# ------------------------------------------------------------
# CryoSleep Imputation
# ------------------------------------------------------------

mask = train["CryoSleep"].isna() & (train["TotalSpend"] > 0)
train.loc[mask, "CryoSleep"] = False

mask = test["CryoSleep"].isna() & (test["TotalSpend"] > 0)
test.loc[mask, "CryoSleep"] = False

mode_cryo = train["CryoSleep"].mode()[0]

train["CryoSleep"].fillna(mode_cryo, inplace=True)
test["CryoSleep"].fillna(mode_cryo, inplace=True)

# Suggested Feature Addition: CryoSpendMismatch & 5. CryoNoSpend
for df in [train, test]:
    df["CryoSpendMismatch"] = (
        (df["CryoSleep"] == True) &
        (df["TotalSpend"] > 0)
    ).astype(int)
    df["CryoNoSpend"] = (
        (df["CryoSleep"] == True) &
        (df["TotalSpend"] == 0)
    ).astype(int)

# ------------------------------------------------------------
# VIP Imputation
# ------------------------------------------------------------

mode_vip = train["VIP"].mode()[0]

train["VIP"].fillna(mode_vip, inplace=True)
test["VIP"].fillna(mode_vip, inplace=True)

# ------------------------------------------------------------
# Cabin Features
# ------------------------------------------------------------

train["Deck"].fillna("Unknown", inplace=True)
test["Deck"].fillna("Unknown", inplace=True)

train["Side"].fillna("Unknown", inplace=True)
test["Side"].fillna("Unknown", inplace=True)

median_cabin = train["CabinNum"].median()

train["CabinNum"].fillna(median_cabin, inplace=True)
test["CabinNum"].fillna(median_cabin, inplace=True)

# Suggested Feature Addition: DeckSide
for df in [train, test]:
    df["DeckSide"] = (
        df["Deck"].astype(str) + "_" + df["Side"].astype(str)
    )

# Suggested Feature Addition: CabinBin & 4. Cabin Position
for df in [train, test]:
    df["CabinBin"] = (df["CabinNum"] // 100).astype(int)
    df["CabinFront"] = (df["CabinNum"] < 300).astype(int)
    df["CabinMiddle"] = (
        (df["CabinNum"] >= 300) &
        (df["CabinNum"] < 700)
    ).astype(int)
    df["CabinRear"] = (df["CabinNum"] >= 700).astype(int)


# ------------------------------------------------------------
# Drop Unnecessary Columns
# ------------------------------------------------------------

drop_cols = [
    "PassengerId",
    "Cabin",
    "Name"
]

train.drop(columns=drop_cols, inplace=True, errors="ignore")
test.drop(columns=drop_cols, inplace=True, errors="ignore")

# ------------------------------------------------------------
# Final Validation
# ------------------------------------------------------------

print("=" * 60)
print("Missing Values")

print("\nTrain")
print(train.isnull().sum()[train.isnull().sum() > 0])

print("\nTest")
print(test.isnull().sum()[test.isnull().sum() > 0])

print("=" * 60)

print("Processed Train Shape :", train.shape)
print("Processed Test Shape  :", test.shape)

display(train.head())

# ------------------------------------------------------------
# Save Processed Data
# ------------------------------------------------------------

train.to_csv("processed_train.csv", index=False)
test.to_csv("processed_test.csv", index=False)
```

```python
# ============================================================
# LOGISTIC REGRESSION - COMPLETE PIPELINE
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    cross_val_score
)

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    RocCurveDisplay
)

# ============================================================
# Create Copies
# ============================================================

train_processed = train.copy()
test_processed = test.copy()

# ============================================================
# Separate Features and Target
# ============================================================

y = train_processed["Transported"].astype(int)

X_train_features = train_processed.drop(
    columns=["Transported"]
)

# ============================================================
# One-Hot Encoding
# ============================================================

categorical_cols = X_train_features.select_dtypes(
    include=["object", "bool", "category"]
).columns.tolist()

X_train_encoded = pd.get_dummies(
    X_train_features,
    columns=categorical_cols,
    drop_first=True
)

X_test_encoded = pd.get_dummies(
    test_processed,
    columns=categorical_cols,
    drop_first=True
)

X, X_test_final = X_train_encoded.align(
    X_test_encoded,
    join="left",
    axis=1,
    fill_value=0
)

# ============================================================
# Train Validation Split
# ============================================================

X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# ============================================================
# Feature Scaling
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_valid_scaled = scaler.transform(X_valid)

# ============================================================
# Logistic Regression Model
# ============================================================

lr_model = LogisticRegression(
    random_state=42,
    max_iter=2000,
    solver="lbfgs",
    C=1.0
)

lr_model.fit(
    X_train_scaled,
    y_train
)

# ============================================================
# Cross Validation
# ============================================================

cv_scores = cross_val_score(
    LogisticRegression(
        random_state=42,
        max_iter=2000,
        solver="lbfgs",
        C=1.0
    ),
    X_train_scaled,
    y_train,
    cv=5,
    scoring="accuracy"
)

# ============================================================
# Prediction
# ============================================================

y_pred = lr_model.predict(X_valid_scaled)

y_prob = lr_model.predict_proba(
    X_valid_scaled
)[:, 1]

# ============================================================
# Evaluation
# ============================================================

accuracy = accuracy_score(y_valid, y_pred)
precision = precision_score(y_valid, y_pred)
recall = recall_score(y_valid, y_pred)
f1 = f1_score(y_valid, y_pred)
roc_auc = roc_auc_score(y_valid, y_prob)

print("="*70)
print("LOGISTIC REGRESSION PERFORMANCE")
print("="*70)

print(f"Accuracy       : {accuracy:.4f}")
print(f"Precision      : {precision:.4f}")
print(f"Recall         : {recall:.4f}")
print(f"F1 Score       : {f1:.4f}")
print(f"ROC-AUC Score  : {roc_auc:.4f}")

print("\nCross Validation Accuracy")

print(cv_scores)

print(f"Mean CV Accuracy : {cv_scores.mean():.4f}")
print(f"Std CV Accuracy  : {cv_scores.std():.4f}")

print("\nClassification Report\n")

print(classification_report(
    y_valid,
    y_pred
))

# ============================================================
# Confusion Matrix
# ============================================================

cm = confusion_matrix(
    y_valid,
    y_pred
)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not Transported","Transported"],
    yticklabels=["Not Transported","Transported"]
)

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.title("Confusion Matrix")

plt.show()

# ============================================================
# ROC Curve
# ============================================================

RocCurveDisplay.from_predictions(
    y_valid,
    y_prob
)

plt.title("ROC Curve - Logistic Regression")

plt.show()

# ============================================================
# Feature Importance
# ============================================================

importance = pd.DataFrame({

    "Feature": X.columns,

    "Coefficient": lr_model.coef_[0],

    "Absolute Importance": np.abs(
        lr_model.coef_[0]
    )

})

importance = importance.sort_values(
    by="Absolute Importance",
    ascending=False
)

print("\nTop 20 Important Features\n")

display(
    importance.head(20)
)

# ============================================================
# Feature Importance Plot
# ============================================================

plt.figure(figsize=(10,8))

sns.barplot(
    data=importance.head(20),
    x="Absolute Importance",
    y="Feature",
    palette="viridis"
)

plt.title("Top 20 Important Features")

plt.xlabel("Absolute Coefficient")

plt.ylabel("Feature")

plt.show()
```

```python
# ============================================================
# MODEL PERFORMANCE COMPARISON
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Import necessary model classes and scaler for retraining
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from sklearn.preprocessing import StandardScaler

# Retrain models with the current X_train and y_train to resolve feature mismatch
# This ensures models are aligned with the latest feature engineering.

# Logistic Regression
scaler = StandardScaler() # Re-initialize scaler with latest X_train
X_train_scaled = scaler.fit_transform(X_train)
X_valid_scaled = scaler.transform(X_valid)

lr_model = LogisticRegression(
    random_state=42,
    max_iter=2000,
    solver="lbfgs",
    C=1.0
)
lr_model.fit(X_train_scaled, y_train)

# Random Forest
rf_model = RandomForestClassifier(
    n_estimators=500,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    bootstrap=True,
    oob_score=True,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train, y_train)

# XGBoost
xgb_model = XGBClassifier(
    n_estimators=500,
    learning_rate=0.03,
    max_depth=5,
    min_child_weight=2,
    subsample=0.8,
    colsample_bytree=0.8,
    gamma=0,
    reg_alpha=0,
    reg_lambda=1,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1,
    tree_method="hist",
    early_stopping_rounds=30
)
# Note: eval_set needs to be consistent with the features used for training
xgb_model.fit(
    X_train,
    y_train,
    eval_set=[(X_valid, y_valid)],
    verbose=False
)

# CatBoost
cat_model = CatBoostClassifier(
    iterations=500,
    learning_rate=0.05,
    depth=6,
    loss_function="Logloss",
    eval_metric="Accuracy",
    random_seed=42,
    verbose=0
)
cat_model.fit(X_train, y_train)


models = {
    "Logistic Regression": (lr_model, X_valid_scaled),
    "Random Forest": (rf_model, X_valid),
    "XGBoost": (xgb_model, X_valid),
    "CatBoost": (cat_model, X_valid)
}

results = []

for model_name, (model, X_eval) in models.items():

    y_pred = model.predict(X_eval)

    # CatBoost returns float/object sometimes
    if model_name == "CatBoost":
        y_pred = y_pred.astype(int)

    y_prob = model.predict_proba(X_eval)[:, 1]

    results.append({

        "Model": model_name,

        "Accuracy": accuracy_score(y_valid, y_pred),

        "Precision": precision_score(y_valid, y_pred),

        "Recall": recall_score(y_valid, y_pred),

        "F1 Score": f1_score(y_valid, y_pred),

        "ROC AUC": roc_auc_score(y_valid, y_prob)

    })

comparison = pd.DataFrame(results)

comparison = comparison.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

comparison.iloc[:,1:] = comparison.iloc[:,1:].round(4)

print("="*80)
print("MODEL PERFORMANCE COMPARISON")
print("="*80)

display(comparison)
```
